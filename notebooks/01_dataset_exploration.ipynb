## 5. Questions for the Learner to Investigate

Once you have extracted the dataset and executed the inspection above, answer the following questions before proceeding to annotation parsing:

1. **Directory Structure & Splits:**
   - Which splits are present (e.g. `train`, `val`, `test`)?
   - Are image files and annotations kept in separate folders or co-located?

2. **Annotation File Discovery:**
   - What are the names and extensions of the candidate annotation files found?
   - Open one of the candidate annotation files in a text editor. Is it a COCO-style JSON file, YOLO text files, Pascal VOC XMLs, or a custom format?

3. **Target Classes:**
   - What classes are annotated (e.g., `swimmer`, `floater`, `boat`, `life jacket`)?
   - Are there distinctions between swimmers in distress vs. recreational swimmers in the annotations?

4. **Image Resolutions & Target Scales:**
   - What are the dimensions (width and height) of the captured UAV images?
   - How small are swimmer bounding boxes compared to the overall frame resolution?

5. **Environmental Factors:**
   - What visual challenges are apparent in the images (e.g. sun glint, wave crests, camera angle, flight altitude)?
